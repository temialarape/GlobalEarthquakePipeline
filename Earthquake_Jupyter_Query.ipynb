```python
import pandas as pd
import numpy as np
import datetime
```

```python
df = pd.read_csv(r"C:\Users\Quitech Info\earthquakes_usgs.csv", header= 0,)
df.head()
```

```python
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])
df.dtypes
```

```python
df['month_name'] = df['date'].dt.month_name()

df.dtypes
```

```python

df.info()
```

#### Top 10 Strongest Earthquakes

```python
df_strongest_earthquakes = df.sort_values('magnitude', ascending=False).head(10)

df_strongest_earthquakes
```

#### Earthquakes Grouped By Month

```python
df['month_number'] = pd.to_datetime(df['date']).dt.month

df.head(2)
```

```python
df_earthquakesByMonth = df.groupby(['month_number', 'month_name']).size()

df_earthquakesByMonth
```

#### How Many Earthquakes Had A Tsunami Flag?

```python
df_earthquakesTsunami = df['tsunami'].value_counts()

df_earthquakesTsunami
```

#### Top 10 Locations With Most Earthquakes

```python
df_earthquakestoploc = df['Location'].value_counts().head(10)

df_earthquakestoploc
```

#### Top 10 deepest earthquakes

```python
df_earthquakesdeep = df.sort_values('depth_km', ascending=False).head(10)

df_earthquakesdeep
```

```python
df_deep = df.sort_values('depth_km', ascending=False).head(10)

df_deep[['date', 'Location', 'depth_km']]
```

#### Is there a correlation between magnitude and depth?

```python
df_corr = df[['magnitude', 'depth_km']].corr()

df_corr
```

#### Earthquakes Grouped by Severity Category

```python
df_severity_category = df.groupby('severity_category').size()

df_severity_category
```

#### Locations where felt reports were recorded

```python
df_felt_reports = df.groupby('Location')['felt_reports'].sum().sort_values(ascending=False).head(10)

df_felt_reports
```
